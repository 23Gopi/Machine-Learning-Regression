```python
import pickle
filename="finalized SVRM_Preboth_model.sav"
scaler_input_filename="finalized inputs.sav"
scaler_output_filename="finalized output.sav"
```

```python
age_input=float(input("Age:"))
bmi_input=float(input("BMI"))
childer_input=float(input("Chilrend "))
gender_input=int(input("Enter the Gender 0 or 1"))
new_input=int(input("Enter smoker 0 or 1:"))
```

```python
scaler_input_model = pickle.load(open(scaler_input_filename,'rb'))
preinput = scaler_input_model.transform([[age_input, bmi_input, childer_input,gender_input,new_input]])
preinput
```

```python
loaded_model=pickle.load(open(filename,'rb'))
result=loaded_model.predict(preinput)
result
```

```python
scaler_output_model = pickle.load(open(scaler_output_filename,'rb'))
preoutput=scaler_output_model.inverse_transform([result])
preoutput
```
