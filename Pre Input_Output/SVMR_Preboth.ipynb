```python
import pandas as pd
```

```python
dataset = pd.read_csv("insurance_pre.csv")
dataset
```

```python

dataset = pd.get_dummies(dataset, drop_first=True)
dataset
```

```python
dataset.columns
```

```python
independent=dataset[['age', 'bmi', 'children', 'sex_male', 'smoker_yes']]
dependent=dataset[['charges']]
```

```python
#split into training set and test
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(independent, dependent, test_size = 0.30, random_state = 0)
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
#fit_transform means it will find miue and standard diviation value(refer note for miue and standard divivation symbom)
X_train=sc.fit_transform(X_train)
#Here we don't use fit because it use the miue and standard diviation value from X_train
X_test=sc.transform(X_test)
```

```python
scy=StandardScaler()
y_train=scy.fit_transform(y_train)
y_test=scy.transform(y_test)
```

```python
from sklearn.svm import SVR
regressor=SVR(kernel="rbf", C=1000)  
regressor.fit(X_train, y_train)
```

```python
regressor.intercept_
```

```python
regressor.n_support_
```

```python
regressor.support_
```

```python
y_pred=regressor.predict(X_test)
from sklearn.metrics import r2_score
r_score=r2_score(y_test,y_pred)
r_score
```

```python
filename="finalized SVRM_Preboth_model.sav"
scaler_input_filename="finalized inputs.sav"
scaler_output_filename="finalized output.sav"
```

```python
pickle.dump(regressor, open(filename,'wb'))
pickle.dump(sc,open(scaler_input_filename,'wb'))
pickle.dump(scy,open(scaler_output_filename,'wb'))
```
